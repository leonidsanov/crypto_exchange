API-ключ — это уникальный идентификатор, который позволяет вам взаимодействовать с API биржи, такой как Bybit. С помощью API-ключа вы можете автоматизировать торговлю, получать рыночные данные и управлять своим аккаунтом.

### Как создать API-ключ на Bybit
Вот пошаговая инструкция по созданию API-ключа на Bybit:

1. **Войдите в свой аккаунт Bybit**: Перейдите на сайт Bybit и выполните вход в свой аккаунт. Если у вас нет аккаунта, зарегистрируйтесь.
2. **Перейдите в настройки API**: В правом верхнем углу страницы нажмите на значок с изображением человека и выберите "Управление API"¹.
3. **Создайте новый API-ключ**: Нажмите "Создать новый ключ" на странице управления API.
4. **Настройте API-ключ**: В появившемся окне настройте свой API-ключ в соответствии с вашими потребностями. Выберите разрешения, которые хотите предоставить (например, чтение данных, торговля и т.д.).
5. **Пройдите проверку безопасности**: Введите код из Google Authenticator для двухфакторной аутентификации (2FA) и нажмите "Следующий шаг".
6. **Сохраните ключи**: После успешного создания API-ключа сохраните API Key и Secret Key в безопасном месте. Эти ключи понадобятся для взаимодействия с API.

```python
import requests
```

```python
# URL для получения списка символов
url = "https://api.bybit.com/v2/public/symbols"
```

```python
# Выполнение GET-запроса
response = requests.get(url)
```

```python
# Проверка успешности запроса
if response.status_code == 200:
    data = response.json()
    symbols = data['result']
    for symbol in symbols:
        print(f"Symbol: {symbol['name']}, Base Currency: {symbol['base_currency']}, Quote Currency: {symbol['quote_currency']}")
else:
    print(f"Failed to fetch symbols: {response.status_code}")
```

```python
# Получение информации о всех инструментах
instruments_info = session.get_instruments_info(category="linear")
```

```python
# Проверка успешности запроса
if 'result' in instruments_info:
    instruments = instruments_info['result']['list']
    
    # Создание DataFrame из списка инструментов
    df = pd.DataFrame(instruments)
    
    # Настройка доступа к Google Sheets
    scope = ["https://spreadsheets.google.com/feeds", "https://www.googleapis.com/auth/drive"]
    creds = ServiceAccountCredentials.from_json_keyfile_name("path/to/your/credentials.json", scope)
    client = gspread.authorize(creds)
    
    # Открытие Google Таблицы (создайте новую таблицу и укажите ее название)
    spreadsheet = client.open("Bybit Instruments")
    
    # Выбор листа (создайте новый лист или используйте существующий)
    worksheet = spreadsheet.sheet1
    
    # Очистка листа перед записью новых данных
    worksheet.clear()
    
    # Запись данных DataFrame в Google Таблицу
    worksheet.update([df.columns.values.tolist()] + df.values.tolist())
    
    print("DataFrame успешно сохранен в Google Таблицу")
else:
    print(f"Failed to fetch instruments: {instruments_info}")
```
